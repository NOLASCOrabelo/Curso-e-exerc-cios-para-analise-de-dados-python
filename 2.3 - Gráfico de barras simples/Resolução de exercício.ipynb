```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/0204_grafico_de_barras_com_duas_ou_mais_variaveis.csv')
```

```python
df.head()
```

```python
df.sort_values(by='Award', ascending=True, inplace=True)
```

```python
df.plot(
   x='Film',
   y=['Award', 'Nomination'],
   ylabel='Quantidade de premiações',
   color=['Green', 'Red'],
   kind='bar',
   title='Oscar 2022, filmes mais votados!'
)
plt.show()
```
