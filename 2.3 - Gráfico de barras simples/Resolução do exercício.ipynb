```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/0205_grafico_de_linhas.csv')
```

```python
df.sample(5)
```

```python
taxa_recife = df[df['municipio'] == 'Recife'] ['taxa_liquida_matricula_pre_escola'].to_list()
taxa_recife
```

```python
anos = df[df['municipio'] =='Recife'] ['ano'].to_list()
```

```python
fig,ax = plt.subplots(figsize=(10,5))
ax.plot(anos,taxa_recife, marker='o')
ax.set(xlabel= 'Anos', ylabel='Taxa')
ax.set_title('Taxa de Matrícula Pre-Escola em Recife', pad=15)
plt.show()
```

```python
municipios = df['municipio'].unique()
```

```python
valores_municipios = ['sao_luis', 'teeresina', 'fortaleza', 'natal', 'joao_pessoa', 'recife', 'maceio', 'aracaju', 'salvador']
```

```python
ano= df['ano'].unique()
```

```python
fig, ax = plt.subplots(figsize=(15, 7), layout='constrained')
for municipio, valor_municipio in zip(municipios, valores_municipios):
    taxa = df[df['municipio'] == municipio]['taxa_liquida_matricula_pre_escola'].to_list()
    ax.plot(ano, taxa, linewidth=2.2, marker='D', label=municipio)

ax.set_title('Taxa de escolarização liquída da pré-escola 2010 e 2019 nos estados do Nordeste', pad=15)
ax.set_xlabel('Ano')
ax.set_ylabel('Taxa')
ax.legend()
plt.show()
```
