```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = df = pd.read_csv('/content/0206_grafico_de_barras_linha.csv')
```

```python
df.sample(3)
```

```python
fig,ax = plt.subplots()
ax.bar(df['mes'],df['meta'], label='Meta', color='Gray')
ax.plot(df['mes'],df['faturamento'], label='Faturamento', linewidth=2, color='Blue')
ax_faturamento = ax.secondary_yaxis('right')
ax.set(ylabel='Valor em R$')
ax.set(xlabel='Mês')
ax.set_title('Faturamento por mês')
ax.legend()
plt.xticks(rotation=45)

plt.show()
```
