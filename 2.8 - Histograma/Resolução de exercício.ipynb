```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv('/content/0208_histograma.csv')
```

```python
df.head()
```

```python
plt.figure(figsize=(7,4))
plt.title('Iris - largura da sepála')
plt.xlabel('Largura de sépala')
plt.ylabel('Frequência')

sns.histplot(data=df, x='SepalLengthCm',color='Red' )


plt.show()
```
